## 2.1: A geometric way of thinking

Consider the one-dimensional dynamical system

$$ \dot{x} = \sin{x} $$

$ $with $x = x_0$ at $t = 0$

This happens to be exactly solvable, *but* the solution is difficult to interpret...

$$ t = \ln{\frac{\csc{x_0}+\cot{x_0}}{\csc{x}+\cot{x}}} $$


#### Example

Below is a python script producing a plot for this function with arbitrary $x_0$, use this to investigate the questions:

$ $1) For $x_0 = \pi/4$, what are the features of $x(t)$ for all $t > 0$? What happens as $t \rightarrow \infty$?

$ $2) For an arbitrary $x_0$, what happens as $t \rightarrow \infty$?


```python
# TASK: investigate the above questions using this script. 
# Here python's built-in ODE solver is used to produce a simple plot of the solution for a given x0

# import useful packages
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt

# differential equation defining system
def flow(x, t):
    dxdt = np.sin(x)
    return dxdt

# user to select initial value x0 here
x0 = np.pi/4

# set up time points
t = np.arange(0,6,0.1)

# solve!
x = odeint(flow, x0, t)

# plot solution
plt.plot(t, x)
plt.xlabel('t')
plt.ylabel('x')
```

```python
# TASK: the picture becomes clearer when we consider multiple different x_0 values at once

import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt

# differential equation defining system
def flow(x, t):
    dxdt = np.sin(x)
    return dxdt

# user to select initial vector of x0 values here.
# try increasing the range, and reducing the spacing
x0 = np.linspace(-np.pi,np.pi,5)

# set up time points
t = np.arange(0,6,0.1)

# solve!
x = odeint(flow, x0, t)

# plot solution
plt.plot(t, x)
plt.xlim(0,6)
plt.xlabel('t')
plt.ylabel('x')
```

i.e. we can achieve a physical understanding of the system without solving $x(t)$.

Think of t as time, x as position of an imaginary particle moving along the real line and $\dot{x}$ as the velocity of that particle.

Hence, the differential equation $\dot{x} = \sin{x}$ represents a *vector field* on the line.


*A more physical picture...* Think of a fluid flowing along the x axis with a velocity that varies from place to place as $\dot{x} = \sin{x}$.


$\bullet$ Flow is to the right when $\dot{x} > 0$; to the left when $\dot{x}<0$

$\bullet$ Points where $\dot{x} = 0$ are known as *fixed points*

$\bullet \equiv$ stable fixed points (or 'attractors' or 'sinks'...)

$\circ \equiv$ unstable fixed points (or 'repellers' or 'sources')

Now we can answer the questions in the previous section by starting our imaginary particle at $x_0$ and watching how it is carried along by the flow.

#### Answers

$ $1) - Particle starting at $x_0 = \pi/4$ moves to the right faster and faster until it crosses $x = \pi/2$ (where sinx is a maximum). The particle then starts slowing down and eventually approaches the stable fixed point at $x = \pi$ from the left.

### 2.2 Fixed points and Stability

$ $Consider a general 1-dimensional system $\dot{x} = f(x)$. Note *f* is independent of $t$.

<div>
<img src="figures/Fig2_2_1.png" width="500"/>
</div>

$\bullet$ To *find the solution to* $\dot{x} = f(x)$ given $x = x_0$ at $t = 0$ we place an imaginary particle (a phase point) at $x_0$ and watch how it is carried along by the fluid flow.


$\bullet$ As the imaginary particle moves along the x axis, its *trajectory* (position as a function of time) is given by x(t).

$\bullet$ A picture which shows all the qualitatively different trajectories of the system is called the *phase portrait*, e.g. the figure we used for our graphical analysis of sin x

$\bullet$ The appearance of the phase portrait is controlled by the fixed points $x^*$ defined by $f(x^*) = 0$

$\bullet$ Fixed points represent (stable or unstable) equilibrium solutions of $\dot{x} = f(x)$

#### Example: Population Growth

Verhulst (1838) suggested the logistic equation

$$ \dot{N} = rN\left(1 - \frac{N}{K}\right) $$

where N(t) is the population at time t, r is the growth rate (r > 0) and K represents the carrying capacity of the population


$\bullet$ $ K \rightarrow \infty \Rightarrow N(t) \propto e^{rt} $ hence population keps growing!

$\bullet$ Finite K models the effect of overcrowding and limited resources

For the logistic equation,

1) Find all the fixed points and

2) Classify their stability.

Let's solve this problem graphically

<div>
<img src="figures/fig2_3_1.png" width="600"/>
</div>

So we see a stable equilibrium at N = K: if N is disturbed slightly from K the disturbance decays and $N(T) \rightarrow K$ as $t \rightarrow \infty$

We have an unstable equilibrium at N = 0: small population grows exponentially fast and runs away from N = 0

For N(t = 0) > 0 population always approaches the carrying capacity K!

```python
# We can check our understanding by numerically solving the logistic equation

import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt

# differential equation defining system
def flow(x, t):
    dxdt = r*x*(1-x/K)
    return dxdt

# set up constants
r = 5     # birth rate 
K = 100   # carrying capacity

# user to select initial vector of x0 values here.
# try increasing the range, and reducing the spacing
x0 = np.linspace(1,200,6)

# set up time points
t = np.arange(0,2,0.01)

# solve!
x = odeint(flow, x0, t)

# plot solution
plt.plot(t, x)
plt.xlim(0,2)
plt.xlabel('t')
plt.ylabel('x')
```

### 2.3: Linear Stability Analysis

$ $The rate of decay to a stable fixed point $x^*$ is obtained by linearizing about $x^∗$....


$ $Let $\delta x(t) = x(t) - x^*$ be a *small perturbation* away from $x^*$. Differentiating $\delta x(t)$ gives

$$ \dot{\delta x} = \frac{d}{dt}(x - x^*) = \dot{x} $$

$ $since $x^*$ is constant.



Hence (via a Taylor expansion)

$$ \dot{\delta x} = f(x) = f(x^* + \delta x) $$
$$                = f(x^*) + \delta x f^{\prime}(x^*) + O(\delta x ^2) $$
$$                = \delta x f^{\prime}(x^*) + O(\delta x ^2) $$

$ $since $f(x^*) = 0$, 

thus

$$ \dot{\delta x} \simeq \delta x f^{\prime}(x^*) $$

$ $which is a *linear equation* in $\delta x$

The solutions to $$ \dot{\delta x} \simeq \delta x f^{\prime}(x^*) $$ 

describe exponential growth and decay, with characteristic timescale $|f'(x^*)|^{-1}$.

$ $The slope $f^{\prime}(x^*)$ at the fixed point determines its stability.

$ $$f^{\prime}(x^*) > 0$: fixed point is *unstable* - perturbations grow exponentially at a rate $f^{\prime}(x^*)$

$ $$f^{\prime}(x^*) < 0$: fixed point is *stable* - perturbations shrink exponentially at a rate $f^{\prime}(x^*)$

#### Example: Population Growth

$ $1) Using linear stability analysis, determine stability for the logistic equation model of population growth



#### Answer

$\dot{N} = rN\left(1 - \frac{N}{K}\right)$

$\bullet$ Fixed points: $f(N) = rN\left(1 - \frac{N}{K}\right) = 0 \Rightarrow N^* = 0, K$

$\bullet$ Stability: $f^{\prime}(N) = r - \frac{2rN}{K} = \begin{cases} \mbox{r} & \mbox{N* = 0:} & \mbox{unstable} \\ \mbox{-r} & \mbox{N* = K:} & \mbox{stable} \end{cases}$

with characteristic timescale $1/r$.

i.e. the same as we found using the graphical method.

### 2.4 Existence and Uniqueness

We need to beware of pathological cases ...

E.g. $\dot{x} = x^{1/3}$ with $x(0) = 0$.


<div>
<img src="figures/Fig2_5_1.png" width="400"/>
</div>





Fixed point: $x^* = 0$

But we can find another solution!

$x(t) = (2t/3)^{3/2}$

Taken together these give infinitely many solutions starting from $x(0) = 0$.

The vector field shows that $x^*=0$ is *very* unstable as $f'(0) \rightarrow \infty$.


#### Existence and Uniqueness Theorem

$\bullet$ Consider $\dot{x} = f(x)$ with $x(0) = 0$.

$\bullet$ Suppose that $f(x)$ and $f'(x)$ are continuous on an open interval $R$ of the $x$-axis and that $x_0$ is a point in $R$.

$\bullet$ Then the initial value problem has a solution $x(t)$ on some time interval $(−\tau, \tau )$ about $t = 0$ and the solution is unique.


In other words ...

'If $f(x)$ is smooth enough, then solutions exist and are unique'


#### Example

$\dot{x} = f(x) = 1+x^2$ with $x(0) = x_0$

$\bullet$ $f(x)$ and $f'(x)$ are continuous for all $x$, so solutions exist and are unique for any $x_0$.

$\bullet$ *but* they do not necessarily exist for all time ... !?

For example, consider $x_0 = 0 \rightarrow x(t) =$ tan $t$




This solution exists only for $- \pi/2 \lt t \lt \pi/2$ because $x(t) \rightarrow \pm \infty$ as $t \rightarrow \pm \pi/2$.

Thus for $|t| \gt \pi/2$, no solution exists given $x_0 = 0$

### 2.5 The Impossibility of Oscillations

$ $There are no periodic solutions to the equation $\dot{x} = f(x)$ since $\dot{x} = f(x)$ corresponds to flow on a line.

Hence, “you never come back to your starting place!”
