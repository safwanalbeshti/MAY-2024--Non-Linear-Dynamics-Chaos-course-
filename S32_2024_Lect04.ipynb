## 4.1 Linear Systems in 2D

In higher dimensions, trajectories have more room to manoeuvre, and hence a wider range of behaviour is possible.

A 2-dimensional linear system has the form

$$ \dot{x} = ax + by $$
$$ \dot{y} = cx + dy $$

where $a$, $b$, $c$, $d$ are parameters.

Equivalently, in vector notation

$$ \dot{\mathbf{x}} = \mathbf{A} \mathbf{x}$$

Where

$$ \mathbf{A} = \begin{pmatrix} a & b \\ c & d \end{pmatrix} \quad \mathrm{and} \quad  \mathbf{x} = \begin{pmatrix} x \\ y \end{pmatrix} $$


$ $The linear property means that if $x_1$ and $x_2$ are solutions, then so is $c_1x_1 + c_2x_2$ for any $c_1$ and $c_2$.

The solutions of $\dot{\mathbf{x}} = \mathbf{A}\mathbf{x}$ can be visualized as trajectories moving on the ($x, y$) plane, or phase plane. This is best illustrated by examples.

#### Example: the simple harmonic oscillator

The state of the system is characterized by $x$ and $v = \dot{x}$

$$ \dot{x} = v $$
$$ \dot{v} = -\frac{k}{m}x $$

or $\dot{\mathbf{x}} = \mathbf{A} \mathbf{x}$ where

$$ \mathbf{A} = \begin{pmatrix} 0 & 1 \\ -\frac{k}{m} & 0 \end{pmatrix} \quad \mathrm{and} \quad  \mathbf{x} = \begin{pmatrix} x \\ v \end{pmatrix} $$


As for a 1-dimensional system, we imagine a fluid flowing steadily on the phase plane with a local velocity given by $(\dot{x}, \dot{v}) = (v, -\omega^2x)$, where $\omega^2 = \frac{k}{m}$ 

 $\Rightarrow$ a vector field on the phase plane.

Task: Use the following python script to visualise the *phase portrait* for the system. Recall that a phase portrait sketches the different trajectories of the system.

```python
import sys
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# Grid of x, y points
nx, ny = 64, 64
x = np.linspace(-2, 2, nx)
v = np.linspace(-2, 2, ny)
X, V = np.meshgrid(x, v)

# USER to vary values of k and m
k = 1
m = 1

xdot = V
vdot = -(k/m)*X

fig = plt.figure()
ax = fig.add_subplot(111)

# Plot the streamlines with an appropriate colormap and arrow style
# warmer colours = faster flow.
color = np.sqrt(xdot**2 + vdot**2)
ax.streamplot(x, v, xdot, vdot, color = color, linewidth=1,
              density=1, arrowstyle='->', arrowsize=1.5)
ax.set_aspect('equal','box')
plt.xlabel('x')
plt.ylabel('v')
```

## 4.2 Classification of general linear systems

Consider a general 2 × 2 matrix $\mathbf{A}$ such that $\dot{\mathbf{x}} = \mathbf{A}\mathbf{x}$


To solve: try

$$ \mathbf{x}(t) = \mathbf{v}e^{\lambda t} \quad \mathrm{where \; \textbf{v} \; is \; a \; constant \; vector}. $$
$$ \Rightarrow \lambda \mathbf{v} e^{\lambda t} = \mathbf{A} \mathbf{v} e^{\lambda t} $$
$$ \Rightarrow \mathbf{A}\mathbf{v} = \lambda \mathbf{v} $$



$ $Hence if we obtain the eigenvectors $\mathbf{v}$ and eigenvalues $\lambda$, we will have two independent solutions $\mathbf{x}(t)$, from which the general solution can be written as a weighted sum:

$$ \mathbf{x}(t) = c_1 \mathbf{v}_1 e^{\lambda_1 t} + c_2 \mathbf{v}_2 e^{\lambda_2 t} $$

$ $$\boldsymbol{\mathrm{A}} = \begin{pmatrix} a & b \\ c & d \end{pmatrix}$ has eigenvalues $\lambda_1$ and $\lambda_2$, where 

$$ \lambda_{1,2} = \frac{\tau \pm \sqrt{\tau^2 -4\Delta}}{2} $$

$ $with $\tau = \textrm{trace}(\mathbf{A}) = a + d$, $\Delta = |\mathbf{A}| = ad - bc$  

$\bullet$ Useful check when calculating eigenvalues: $\lambda_1 + \lambda_2 = \tau$ and $\lambda_1\lambda_2 = \Delta$

Alternatively you can solve $|\mathbf{A}-\lambda_i\mathbf{I}| = 0$

#### Example

1) Find the eigenvalues and eigenvectors of the following system of equations:

$\begin{pmatrix} \dot{x} \\ \dot{y} \end{pmatrix} = \begin{pmatrix} 1 & 1 \\ 4 & -2 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix}$

```python
import sys
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# Grid of x, y points
nx, ny = 64, 64
x = np.linspace(-2, 2, nx)
y = np.linspace(-2, 2, ny)
X, Y = np.meshgrid(x, y)

# USER to set equations here
xdot = X + Y
ydot = 4*X - 2*Y

fig = plt.figure()
ax = fig.add_subplot(111)

# Plot the streamlines with an appropriate colormap and arrow style
color = np.sqrt(xdot**2 + ydot**2)
ax.streamplot(x, y, xdot, ydot, color = color, linewidth=1,
              density=1, arrowstyle='->', arrowsize=1.5)
ax.set_aspect('equal','box')
plt.xlabel('x')
plt.ylabel('y')
plt.axvline(c='grey', lw=1)
plt.axhline(c='grey', lw=1)
```

## 4.3 Nonlinear systems

### Fixed points and linearisation

Now consider a nonlinear two-dimensional system

$$ \dot{x} = f(x,y) $$
$$ \dot{y} = g(x,y) $$

$ $Suppose $(x^∗,y^∗)$ is a fixed point. Expand around $(x^∗,y^∗)$ using $\delta x = x − x^∗$ and $\delta y = y − y^∗$.



$$ \dot{\delta x} = \dot{x} = f(x^* + \delta x,y^* + \delta y) $$
$$ = f(x^*,y^*) + \delta x\frac{\partial f}{\partial x} + \delta y\frac{\partial f}{\partial y} + O(\delta x^2,\delta y^2,\delta x \delta y) $$
$$ \simeq \delta x\frac{\partial f}{\partial x} + \delta y\frac{\partial f}{\partial y} $$



similarly

$$ \dot{\delta y} \simeq \delta x\frac{\partial g}{\partial x} + \delta y\frac{\partial g}{\partial y} $$



$ $Hence a small disturbance around $(x^∗,y^∗)$ evolves as

$$ \begin{pmatrix} \dot{\delta x} \\ \dot{\delta y} \end{pmatrix} = \begin{pmatrix} \frac{\partial f}{\partial x} & \frac{\partial f}{\partial y} \\ \frac{\partial g}{\partial x} & \frac{\partial g}{\partial y} \end{pmatrix} \begin{pmatrix} \delta x \\ \delta y \end{pmatrix} $$

$ $where the matrix is known as the Jacobian matrix $\mathbf{J}$ at $(x^∗,y^∗)$, and
is the multivariate equivalent of $f^{\prime}(x∗)$ for 1-D systems.


Now we can evaluate the dynamics of the linearised system close to a F.P. using techniques discussed earlier

- Eigenvalues and vectors indicate how small perturbations grow or decay 


#### Example: Rabbits vs Sheep

An example of the *Lotka-Volterra* model of competition between two species (e.g. rabbits, $x$, and sheep, $y$) grazing the same food supply (grass).

- Each species grows to its carrying capacity in the absence of the other - *logistic growth* with rabbits growing faster

- When species encounter each other, the larger (sheep) has an advantage.

- Conflicts occur at a rate proportional to the size of each population. Conflicts reduce the growth rate of each species (but more for rabbits).

$ $A model encapsulating these properties could be $\begin{cases} \dot{x} = 3x(1 - \frac{x}{3}) - 2xy \\ \dot{y} = 2y(1 - \frac{y}{2}) -xy \end{cases}$

For this system

- find the fixed points
- classify their stability
- sketch the phase portrait


#### Answer

$ $In general, we must obtain fixed points by solving $\dot{x} = 0$ and $\dot{y} = 0$ simultaneously, i.e. at the fixed point:

$$ \dot{x} = 3x(1 - x/3) - 2xy = 0 $$
$$ \dot{y} = 2y(1 - y/2) -xy = 0 $$

yields fixed points (0,0), (0,2), (3,0) and (1,1).

The Jacobian for the system is

$\mathbf{J} = \begin{pmatrix} 3-2x-2y & -2x \\ -y & 2-x-2y \end{pmatrix}$


Evaluate $\mathbf{J}$ at each fixed point to find stability:

$ $(0,0) where $\mathbf{J} = \begin{pmatrix} 3 & 0 \\ 0 & 2 \end{pmatrix} \Rightarrow \lambda = 3,2$


$ $(0,2) where $\mathbf{J} = \begin{pmatrix} -1 & 0 \\ -2 & -2 \end{pmatrix} \Rightarrow \lambda = -1,-2$

$ $(1,1) where $\mathbf{J} = \begin{pmatrix} -1 & -2 \\ -1 & -1 \end{pmatrix} \Rightarrow \lambda = -1,\pm\sqrt{2}$

$ $(3,0) where $\mathbf{J} = \begin{pmatrix} -3 & -6 \\ 0 & -1 \end{pmatrix} \Rightarrow \lambda = -3,-1$





$\bullet$ $ $(0,0): $\lambda = 3,2 \Rightarrow$ unstable node (repellor)

$$ \lambda = 2 \Rightarrow \mathbf{v} = (0,1) \quad \textrm{"slow eigendirection"} $$
$$ \lambda = 3 \Rightarrow \mathbf{v} = (1,0) \quad \textrm{"fast eigendirection"} $$

General rule...

$ $Trajectories are tangential to the slow eigendirection (i.e. smallest $|\lambda|$) at a node

#### Example: The simple pendulum

Consider a simple, nondimensional pendulum:

$$ \ddot{\theta} + \sin{\theta} = 0$$

Which can be broken down into 
$$ \dot{\theta} = \omega , \quad \dot{\omega} = -\sin{\theta}  $$

Task: Find the fixed points, evaluate their stability, and sketch the phase portrait.


#### Answer: 

System: $\dot{\theta} = \omega , \, \dot{\omega} = -\sin{\theta}  $

It is helpful to relate the different trajectories in the phase portrait to qualitatively different behaviours of a pendulum.

Task: Describe the possible motions of a simple pendulum using the phase portrait. You might find the python script below helpful.

```python
# TASK: explore different theta0, omega0 to relate phase portrait to movement of pendulum
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt

def flow(x, t):
    theta, omega = x
    dthetadt = omega
    domegadt = -np.sin(theta)  
    return [dthetadt, domegadt]

# user to select initial value x0 here
theta0 = np.pi/4
omega0 = 0

x0 = [theta0,omega0]
t = np.arange(0,10,0.1)
x = odeint(flow, x0, t)

ax2 = plt.subplot(1,2,1)
plt.plot(x[:,0], x[:,1])
plt.xlabel('theta'), plt.ylabel('omega')
plt.title('phase portrait')

ax1 = plt.subplot(1,2,2)
plt.plot(t, x[:,0]), plt.xlabel('t')
plt.title('theta timeseries')
```

#### Example: The simple damped pendulum

Consider the addition of a linear damping term

$$ \dot{\theta} = \omega , \quad \dot{\omega} = -\sin{\theta} -r\omega  $$


The fixed points are unchanged from the undamped case. But the stability properties of the *even* fixed points are modified:

$\lambda \approx \begin{cases} \pm i - \frac{r}{2} \quad (n \, \mathrm{even}) \\ \pm 1 \quad \quad (n \, \mathrm{odd}) \end{cases}$

Assuming weak damping ($r << 1$).

This means the even fixed points are now stable spirals (decaying orbits). All trajectories eventually end up at one of the stable fixed points.


<div>
<img src="figures/4_dampedpendulum.png" width="800"/>
</div>



```python
# TASK: explore different theta0, omega0 to relate phase portrait to movement of a *damped* pendulum
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt

def flow(x, t):
    theta, omega = x
    dthetadt = omega
    domegadt = -np.sin(theta)  - r*omega
    return [dthetadt, domegadt]

# user to select initial value x0 here
theta0 = np.pi/4
omega0 = 0

r = 0.1
x0 = [theta0,omega0]
t = np.arange(0,100,0.1)
x = odeint(flow, x0, t)

ax2 = plt.subplot(1,2,1)
plt.plot(x[:,0], x[:,1])
plt.xlabel('theta'), plt.ylabel('omega')
plt.title('phase portrait')

ax1 = plt.subplot(1,2,2)
plt.plot(t, x[:,0]), plt.xlabel('t')
plt.title('timeseries')
```
